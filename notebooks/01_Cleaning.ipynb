```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'../data\raw\spy_2020_2022.csv', low_memory = False)
```

```python
df.shape
```

```python
df.sample(5).T
```

```python
df.columns = df.columns.str.strip('[ ]').str.lower()
df.columns
```

```python
c_parts = df['c_size'].str.split('x', expand = True)
df['c_size_bid'] = c_parts[0]
df['c_size_ask'] = c_parts[1]
```

```python
p_parts = df['p_size'].str.split('x', expand = True)
df['p_size_bid'] = p_parts[0]
df['p_size_ask'] = p_parts[1]
```

```python
df[['c_size_bid', 'c_size_ask', 'c_size']].sample(5)
```

```python
df[['p_size_bid', 'p_size_ask', 'p_size']].sample(5)
```

```python
df = df.drop(columns = ['c_size', 'p_size'])
```

```python
df.info()
```

```python
to_num = ['c_delta',
       'c_gamma', 'c_vega', 'c_theta', 'c_rho', 'c_iv', 'c_volume', 'c_last',
       'c_size_bid', 'c_size_ask', 'c_bid', 'c_ask', 'strike', 'p_bid', 'p_ask', 'p_size_bid','p_size_ask',
       'p_last', 'p_delta', 'p_gamma', 'p_vega', 'p_theta', 'p_rho', 'p_iv',
       'p_volume', 'strike_distance', 'strike_distance_pct']
```

```python
for i in to_num :
    df[i] = pd.to_numeric(df[i], errors = 'coerce')
```

```python
df.columns
```

```python
df[['quote_unixtime', 'quote_readtime', 'quote_date', 'quote_time_hours',
       'underlying_last', 'expire_date', 'expire_unix', 'dte']].head()
```

```python
to_time = ['quote_readtime', 'quote_date', 'expire_date']
```

```python
for i in to_time :
    df[i] = pd.to_datetime(df[i])
```

```python
df[to_time].info()
```

```python
df[['quote_unixtime', 'quote_readtime', 'quote_date', 'quote_time_hours',
       'underlying_last', 'expire_date', 'expire_unix', 'dte']].head()
```

```python
holidays = ['2022-01-17',
'2022-02-21',
'2022-04-15',
'2022-05-30',
'2022-06-20',
'2022-07-04',
'2022-09-05',
'2022-11-24',
'2022-12-26']
```

```python
holidays = pd.to_datetime(holidays)
```

```python
df.columns
```

```python
df.shape
```

```python
df = df[~df['quote_date'].isin(holidays)]
```

```python
df.shape
```

```python
df.columns
```

```python
call_filt = ['quote_unixtime', 'quote_readtime', 'quote_date', 'quote_time_hours',
       'underlying_last', 'expire_date', 'expire_unix', 'dte', 'c_delta',
       'c_gamma', 'c_vega', 'c_theta', 'c_rho', 'c_iv', 'c_volume', 'c_last',
       'c_bid', 'c_ask', 'strike', 'strike_distance', 'strike_distance_pct', 'c_size_bid', 'c_size_ask']
```

```python
call = df[call_filt]
call.columns = call.columns.str.removeprefix('c_')
call['type'] = 'call'
```

```python
put_filt = ['quote_unixtime', 'quote_readtime', 'quote_date', 'quote_time_hours',
       'underlying_last', 'expire_date', 'expire_unix', 'dte', 'p_delta',
       'p_gamma', 'p_vega', 'p_theta', 'p_rho', 'p_iv', 'p_volume', 'p_last',
       'p_bid', 'p_ask', 'strike', 'strike_distance', 'strike_distance_pct', 'p_size_bid', 'p_size_ask']
```

```python
put = df[put_filt]
put.columns = put.columns.str.removeprefix('p_')
put['type'] = 'put'
```

```python
call.shape
```

```python
put.shape
```

```python
long_df = pd.concat([call, put], ignore_index = True)
long_df.index.is_unique
```
